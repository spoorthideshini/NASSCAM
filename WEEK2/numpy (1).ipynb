```python
import numpy as np
arr=np.arange(1,10,2)
print(arr)
```

```python
arr=np.linspace(0,1,6)
print(arr)
```

```python
arr=np.arange(1,10).reshape(3,3)
print(arr)
```

```python

print("arr:",arr.flatten())
```

```python
arr=[[1,2,3],[6,7,8]]
print(arr)
print("sum:",np.sum(arr))
print("colsum:",np.sum(arr,axis=0))
print("rowsum:",np.sum(arr,axis=1))
print("avg:",np.mean(arr,axis=1))
print("max value:",np.max(arr))
print("min value",np.min(arr))
print("max value index:",np.argmax(arr))
print("std:",np.std(arr))
```

```python
a=[3,6,8,1,9,3]
sorted_array = np.sort(a)
print("Sorted Array:", sorted_array)
idx = np.where(a > 4)
print("Indices where value > 4:", idx)
print("Values > 4:", a[idx])
```
