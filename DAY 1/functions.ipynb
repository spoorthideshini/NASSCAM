```python
def greet():
    print("Hello world")
greet()
```

```python
def greet(name):
    print(f"Hello, {name}!")

greet("sony")
greet("chintu")
```

```python
def greet(name="guest"):
    print(f"welcome, {name}!")

greet()
greet("sony")
```

```python
def printnames(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

printnames("sony", "chinti")
```

```python
def printinfo(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

printinfo(name="sony", age=18, city="karimnagar")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 16:", square(16))
```
