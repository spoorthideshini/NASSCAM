```python
from datetime import datetime, date, timedelta
now = datetime.now()
print("now:", now)
today = date.today()
print("Today's Date:", today)
dob = date(2008, 2, 19)
print("DOB:", dob)
yesterday = today - timedelta(days=1)
tomorrow = today + timedelta(days=1)
print("Yesterday:", yesterday)
print("Tomorrow:", tomorrow)
formatted = now.strftime("%d-%m-%Y %H:%M:%S")
print("Formatted DateTime:", formatted)
```

```python
import random
print("random integer:", random.randint(1, 1000))
print("random float:", random.random())
color = ['pink', 'black', 'blue']
print("random color:", random.choice(color))
random.shuffle(color)
print("Shuffled List:", color)
print("Sample:", random.sample(range(1, 100), 9))
```

```python
from collections import Counter
fruits = ['apple', 'banana', 'grapes', 'orange', 'kivi', 'watermelon']
count = Counter(fruits)
print("fruits Count:", count)
```

```python
from collections import defaultdict
marks = defaultdict(lambda: 0)
marks['Math'] = 89
marks['Science'] = 36
print("English marks:", marks['English'])
print("math marks:", marks['Math'])
```

```python
from collections import deque
dq = deque([1, 5, 9])
dq.append(8)
dq.appendleft(4)
dq.pop()
dq.popleft()
print("Deque after operations:",dq)
```
